```python
# Question 16: Matrix Multiplication

import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

result = A @ B

print("Matrix A:")
print(A)

print("Matrix B:")
print(B)

print("Matrix Multiplication (A @ B):")
print(result)
```

```python
# Question 17: Max, Min and Index of Max

import numpy as np

arr = np.random.randint(1, 100, 20)

print("Array:", arr)
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
print("Index of Maximum:", np.argmax(arr))
```

```python
# Question 18: Replace Values Less Than 0.5 with 0

import numpy as np

arr = np.random.rand(10)

print("Original Array:")
print(arr)

arr[arr < 0.5] = 0

print("Modified Array:")
print(arr)
```

```python
# Question 19: Clip Values Between 5 and 15

import numpy as np

arr = np.array([2, 5, 7, 10, 16, 20])

clipped = np.clip(arr, 5, 15)

print("Original:", arr)
print("Clipped:", clipped)
```

```python
# Question 21: Boolean Masking

import numpy as np

arr = np.arange(1, 21)

even_numbers = arr[arr % 2 == 0]

print("Original Array:", arr)
print("Even Numbers:", even_numbers)
```

```python
# Question 22: Fancy Indexing

import numpy as np

arr = np.arange(1, 26).reshape(5, 5)

selected_rows = arr[[0, 2, 4]]

print("Original Array:")
print(arr)

print("Rows 0, 2 and 4:")
print(selected_rows)
```

```python
# Question 23: np.where()

import numpy as np

arr = np.arange(1, 21)

labels = np.where(arr > 10, "High", "Low")

print(arr)
print(labels)
```

```python
# Question 24: Unique Elements and Counts

import numpy as np

arr = np.array([1, 2, 2, 3, 3, 3, 4])

unique, counts = np.unique(arr, return_counts=True)

print("Unique Values:", unique)
print("Counts:", counts)
```

```python
# Question 28: Replace NaN with Mean

import numpy as np

arr = np.array([10, 20, np.nan, 40, np.nan, 60])

mean = np.nanmean(arr)

arr[np.isnan(arr)] = mean

print(arr)
```

```python
# Question 29: Sort 2D Array by Second Column

import numpy as np

arr = np.array([
    [5, 9],
    [3, 2],
    [7, 5],
    [1, 8]
])

sorted_arr = arr[arr[:, 1].argsort()]

print("Original Array:")
print(arr)

print("Sorted by Second Column:")
print(sorted_arr)
```
